**Прочность при растяжении, МПа** - Прочность при растяжении (временное сопротивление, предел прочности) (σ_в, МПа) определяется как отношение максимальной нагрузки (Pmax, предшествующей разрушению образца) к начальной площади его поперечного сечения.

График показывает наличие выбросов у каждого признака

```python
def find_outliers_iqr(data, column):
    Q1 = data[column].quantile(0.25)
    Q3 = data[column].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    
    outliers = data[(data[column] < lower_bound) | (data[column] > upper_bound)]
    return outliers, lower_bound, upper_bound

outliers_dict = {}

print("ВЫБРОСЫ ПО КАЖДОМУ ПРИЗНАКУ")

for col in data.columns:
    outliers, lower, upper = find_outliers_iqr(data, col)
    outliers_dict[col] = outliers
    print("\n")
    print(col)
    print(f"Нижняя граница: {lower:.4f}")
    print(f"Верхняя граница: {upper:.4f}")
    print(f"Количество выбросов: {len(outliers)}")
    
    if len(outliers) > 0:
        print(f"Значения выбросов:\n{outliers[col].values}")
        print(f"Индексы выбросов: {outliers.index.tolist()}")
```

Ввиду отсутствия предметных знаний в области композиционных материалов, обоснованность сохранения выбросов неочевидна

Присутствие в выборке образцов с редким составом может негативно сказаться на качестве последующей модели

Рассмотрим признак "Соотношение матрица–наполнитель". Значения 5:1 и 0,3:1 могли бы свидетельствовать об ошибке в данных, поскольку при реальном соотношении 5:1 ожидалось бы сопутствующее отклонение и в других признаках для тех же строк. С учётом этого наблюдения обосновано удаление строк, содержащих выбросы

```python
all_outliers = set()
for col, outliers_df in outliers_dict.items():
    all_outliers.update(outliers_df.index.tolist())

print(f"Найдено уникальных индексов с выбросами: {len(all_outliers)}")
data_clean = data.drop(index=list(all_outliers))
print(f"Было строк: {len(data)}")
print(f"Стало строк: {len(data_clean)}")
```

```python
results = []
best_models = {}

for name in base_models.keys():
    print(f"Модель {name}")
    
    if param_grids[name]: 
        grid = GridSearchCV(
            estimator = base_models[name],
            param_grid = param_grids[name],
            cv = 10,
            scoring='r2',
            n_jobs=-1,
            verbose=1
        )
        
        grid.fit(X_train_scaled, y_train_scaled)
        best_model = grid.best_estimator_
        best_params = grid.best_params_
        cv_score = grid.best_score_
        
    else:  
        best_model = base_models[name]
        best_model.fit(X_train_scaled, y_train_scaled)
        cv_scores = cross_val_score(best_model, X_train_scaled, y_train_scaled, cv=10, scoring='r2')
        cv_score = cv_scores.mean()
        best_params = None
    
    y_pred_scaled = best_model.predict(X_test_scaled)
    y_pred = scaler_y.inverse_transform(y_pred_scaled)

    # метрики
    r2_elastic = r2_score(y_test.iloc[:, 0], y_pred[:, 0])
    r2_strength = r2_score(y_test.iloc[:, 1], y_pred[:, 1])
    mae_elastic = mean_absolute_error(y_test.iloc[:, 0], y_pred[:, 0])
    mae_strength = mean_absolute_error(y_test.iloc[:, 1], y_pred[:, 1])
    rmse_elastic = np.sqrt(mean_squared_error(y_test.iloc[:, 0], y_pred[:, 0]))
    rmse_strength = np.sqrt(mean_squared_error(y_test.iloc[:, 1], y_pred[:, 1]))
    
    best_models[name] = best_model
    
    results.append({
        'Модель': name,
        'R2 CV': cv_score,
        'R2 упругость': r2_elastic,
        'R2 прочность': r2_strength,
        'MAE упругость': mae_elastic,
        'MAE прочность': mae_strength,
        'RMSE упругость': rmse_elastic,
        'RMSE прочность': rmse_strength,
        'Лучшие гипперпараметры': best_params
    })
    
    print(f"R2 CV: {cv_score}")
    print(f"R2 упругость: {r2_elastic}, R2 прочность: {r2_strength}\n\n")
    print(f"MAE упругость: {mae_elastic}, MAE прочность: {mae_strength}\n\n")

results = pd.DataFrame(results).sort_values('MAE упругость', ascending=True)

print("Результаты \n\n")
print(results[['Модель', 'R2 CV', 'R2 упругость', 'R2 прочность', 
                  'MAE упругость', 'MAE прочность',
              'RMSE упругость', 'RMSE прочность']].to_string(index=False))
print(results[['Лучшие гипперпараметры']].to_string(index=False))

best_model_name = results.iloc[0]['Модель']
```

# Нейросеть для рекомендации соотношения матрица-наполнитель #

```python
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Dropout
from tensorflow.keras.optimizers import Adam
from tensorflow.keras.callbacks import EarlyStopping
from tensorflow.keras.regularizers import l2
```

```python
target = [
    'Соотношение матрица-наполнитель'
]

feature = [
    'Модуль упругости при растяжении, ГПа',
    'Прочность при растяжении, МПа',
    'Плотность, кг/м3',
    'модуль упругости, ГПа',
    'Количество отвердителя, м.%',
    'Содержание эпоксидных групп,%_2',
    'Температура вспышки, С_2',
    'Поверхностная плотность, г/м2',
    'Потребление смолы, г/м2',
    'Угол нашивки, град',
    'Шаг нашивки',
    'Плотность нашивки'
]

X = data[feature]
y = data[target]

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=0
)

scaler_X_n = StandardScaler()
X_train_scaled = scaler_X_n.fit_transform(X_train)
X_test_scaled = scaler_X_n.transform(X_test)

scaler_y_n = StandardScaler()
y_train_scaled = scaler_y_n.fit_transform(y_train)
y_test_scaled = scaler_y_n.transform(y_test)
```

```python
model = Sequential([
    Dense(6, activation='relu', kernel_regularizer=l2(0.05), input_shape=(X_train.shape[1],)),
    Dense(3, activation='relu', kernel_regularizer=l2(0.05)),
    Dense(1)
])

model.compile(
    optimizer=Adam(learning_rate=0.001),
    loss='mae',
    metrics=['mae']
)
```

```python
history = model.fit(
    X_train_scaled, y_train_scaled,
    epochs=100,
    batch_size=4,
    validation_data=(X_test_scaled, y_test_scaled),
    callbacks=[EarlyStopping(patience=20, restore_best_weights=True)],
    verbose=1
)
```

```python
y_pred_scaled = model.predict(X_test_scaled, verbose=0).flatten()
```

```python
y_pred = scaler_y_n.inverse_transform(y_pred_scaled.reshape(-1, 1)).flatten()

print("Соотношение матрица-наполнитель")
print(f"MAE:  {mean_absolute_error(y_test, y_pred):.4f}")
print(f"RMSE: {np.sqrt(mean_squared_error(y_test, y_pred)):.4f}")
print(f"R²:   {r2_score(y_test, y_pred):.4f}")
```

```python
y_test_arr = y_test.values.flatten()
y_pred_arr = y_pred.flatten()

plt.plot(range(len(y_test_arr)), y_test_arr, 'o-', label='Реальные', linewidth=1.5)

plt.plot(range(len(y_pred_arr)), y_pred_arr, 's--', label='Предсказанные', linewidth=1.5)

plt.xlabel('Номер образца')
plt.ylabel('Соотношение матрица-наполнитель')
plt.title('Нейронная сеть: реальные vs предсказанные')
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```
